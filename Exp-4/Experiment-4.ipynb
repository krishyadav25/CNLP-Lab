#4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
#Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
#Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
import csv
from collections import Counter
import spacy

# -----------------------------------
# 1. Read input text file
# -----------------------------------

with open("/content/drive/MyDrive/CNLP-Lab/Exp-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# -----------------------------------
# 2. Tokenization and Term Frequency
# -----------------------------------

# Extract words and convert to lowercase
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Count frequency of each word
word_frequency = Counter(words)

# -----------------------------------
# 3. Save Term Frequency to CSV
# -----------------------------------

with open("term_frequency_toolkit.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # Write each term and frequency
    for word, frequency in word_frequency.items():
        writer.writerow([word, frequency])

print("Term frequency saved successfully to term_frequency.csv")

# -----------------------------------
# 4. Display Top 10 Most Frequent Terms
# -----------------------------------

print("\nTop 10 Most Frequent Terms:")
print("----------------------------")

for word, frequency in word_frequency.most_common(10):
    print(f"{word} : {frequency}")

# -----------------------------------
# 5. Named Entity Recognition (NER)
# -----------------------------------

# Load English spaCy model
nlp = spacy.load("en_core_web_sm")

# Process the text
doc = nlp(text)

print("\nNamed Entities:")
print("----------------------------")

for entity in doc.ents:
    print(f"{entity.text} : {entity.label_}")
```

#4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

#Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
#Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
import csv

# -----------------------------------
# 1. Read input text file
# -----------------------------------

with open("/content/drive/MyDrive/CNLP-Lab/Exp-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# -----------------------------------
# 2. Extract words
# -----------------------------------

# Convert text to lowercase
text = text.lower()

# Extract words using regular expression
words = re.findall(r'\b[a-zA-Z]+\b', text)

# -----------------------------------
# 3. Calculate frequency of each word
# -----------------------------------

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# -----------------------------------
# 4. Save frequency to CSV file
# -----------------------------------

with open("term_frequency_native.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # Write terms and frequencies
    for word, count in frequency.items():
        writer.writerow([word, count])

print("Term frequency saved successfully to term_frequency.csv")

# -----------------------------------
# 5. Find Top 10 Most Frequent Terms
# -----------------------------------

# Sort words according to frequency
sorted_words = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

print("\nTop 10 Most Frequent Terms:")
print("----------------------------")

for word, count in sorted_words[:10]:
    print(word, ":", count)
```

#4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
#1.Read multiple text documents.
#2.Convert text to lowercase.
#3.Remove punctuation.
#4.Tokenize the documents.
#5.Calculate TF for every term.
#6.Calculate DF for every term.
#7.Calculate IDF using the logarithmic formula.
#8.Calculate TF-IDF for every term in every document.
#9.Display Term, TF, DF, IDF, and TF-IDF at each step.
#10.Identify the top 10 terms with the highest TF-IDF score for each document.
#Given Input
#Consider the following three documents:
#Document 1:
#Natural language processing is a field of artificial intelligence.

#Document 2:
#Natural language processing helps computers understand human language.

#Document 3:
#Machine learning is an important part of artificial intelligence.


```python
import re
import math

# --------------------------------------------------
# 1. Given Multiple Documents
# --------------------------------------------------

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

# --------------------------------------------------
# 2. Convert to Lowercase and Remove Punctuation
# --------------------------------------------------

cleaned_documents = []

for document in documents:
    # Convert to lowercase
    document = document.lower()

    # Remove punctuation
    document = re.sub(r'[^\w\s]', '', document)

    cleaned_documents.append(document)

# --------------------------------------------------
# 3. Tokenize the Documents
# --------------------------------------------------

tokenized_documents = []

for document in cleaned_documents:
    words = document.split()
    tokenized_documents.append(words)

print("TOKENIZED DOCUMENTS")
print("--------------------")

for i, words in enumerate(tokenized_documents, 1):
    print(f"Document {i}: {words}")

# --------------------------------------------------
# 4. Calculate TF
# TF = Frequency of term / Total number of terms
# --------------------------------------------------

tf_documents = []

for words in tokenized_documents:

    term_count = {}

    # Count each term
    for word in words:
        if word in term_count:
            term_count[word] += 1
        else:
            term_count[word] = 1

    # Calculate TF
    total_words = len(words)
    tf = {}

    for word, count in term_count.items():
        tf[word] = count / total_words

    tf_documents.append(tf)

# --------------------------------------------------
# 5. Calculate DF
# DF = Number of documents containing the term
# --------------------------------------------------

df = {}

for tf in tf_documents:
    for term in tf:
        if term in df:
            df[term] += 1
        else:
            df[term] = 1

# --------------------------------------------------
# 6. Calculate IDF
# IDF = log(N / DF)
# --------------------------------------------------

N = len(documents)

idf = {}

for term, document_frequency in df.items():
    idf[term] = math.log(N / document_frequency)

# --------------------------------------------------
# 7. Calculate TF-IDF
# TF-IDF = TF × IDF
# --------------------------------------------------

tfidf_documents = []

for tf in tf_documents:

    tfidf = {}

    for term, tf_value in tf.items():
        tfidf[term] = tf_value * idf[term]

    tfidf_documents.append(tfidf)

# --------------------------------------------------
# 8. Display Term, TF, DF, IDF and TF-IDF
# --------------------------------------------------

for i in range(N):

    print(f"\n\nDOCUMENT {i + 1}")
    print("=" * 65)

    print(f"{'Term':<20}{'TF':<12}{'DF':<8}{'IDF':<12}{'TF-IDF':<12}")
    print("-" * 65)

    for term in sorted(tf_documents[i].keys()):

        tf_value = tf_documents[i][term]
        df_value = df[term]
        idf_value = idf[term]
        tfidf_value = tfidf_documents[i][term]

        print(
            f"{term:<20}"
            f"{tf_value:<12.4f}"
            f"{df_value:<8}"
            f"{idf_value:<12.4f}"
            f"{tfidf_value:<12.4f}"
        )

# --------------------------------------------------
# 9. Display Top 10 TF-IDF Terms for Each Document
# --------------------------------------------------

for i in range(N):

    print(f"\nTop 10 TF-IDF Terms - Document {i + 1}")
    print("----------------------------------------")

    # Sort terms according to TF-IDF
    sorted_terms = sorted(
        tfidf_documents[i].items(),
        key=lambda x: x[1],
        reverse=True
    )

    for term, score in sorted_terms[:10]:
        print(f"{term:<20} {score:.4f}")
```
